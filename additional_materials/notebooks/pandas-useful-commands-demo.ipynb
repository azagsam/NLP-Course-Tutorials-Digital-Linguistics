This notebook showcases a practical set of pandas commands on a small toy dataset with 10 samples. The goal is to provide a quick reference for the commands that are most useful in day-to-day data exploration and wrangling.

## Import pandas and create toy data

```python
import pandas as pd

data = {
    "student": ["Ana", "Boris", "Cvetka", "David", "Eva", "Filip", "Gaja", "Hana", "Ivan", "Jana"],
    "language": ["Python", "R", "Python", "Java", "Python", "R", "Python", "Java", "Python", "R"],
    "score": [88, 76, 91, 65, 84, None, 95, 72, 89, 78],
    "hours_studied": [12, 8, 15, 6, 10, 7, 16, 9, 13, 11],
    "passed": [True, True, True, False, True, True, True, False, True, True]
}

df = pd.DataFrame(data)
df
```

## Inspect the dataframe

Useful commands: `head()`, `tail()`, `shape`, `columns`, `info()`, `describe()`.

```python
print("Shape:", df.shape)
print("Columns:", list(df.columns))

df.head(3)
```

```python
df.info()
```

```python
df.describe(include="all")
```

## Select columns and filter rows

Useful commands: column selection, boolean filtering, `loc`, `iloc`, `query()`.

```python
# Select two columns
df[["student", "score"]]
```

```python
# Students who studied more than 10 hours and passed
df[(df["hours_studied"] > 10) & (df["passed"])]
```

```python
# loc uses labels, iloc uses integer positions
df.loc[df["language"] == "Python", ["student", "language", "score"]]
```

```python
df.iloc[:4, :3]
```

## Create and modify columns

Useful commands: `assign()`, arithmetic on columns, `map()`, `replace()`.

```python
df2 = df.assign(
    score_filled=df["score"].fillna(df["score"].mean()),
    score_per_hour=(df["score"].fillna(df["score"].mean()) / df["hours_studied"]).round(2),
    level=df["score"].map(lambda x: "missing" if pd.isna(x) else ("high" if x >= 85 else "medium" if x >= 75 else "low"))
)

df2
```

## Sort, count values, and find missing data

Useful commands: `sort_values()`, `value_counts()`, `isna()`, `fillna()`.

```python
df.sort_values(by="score", ascending=False, na_position="last")
```

```python
df["language"].value_counts()
```

```python
print("Any missing values?", df.isna().any().any())
df.isna().sum()
```

## Group by category and aggregate

Useful commands: `groupby()`, `agg()`, `mean()`, `count()`.

```python
df.groupby("language").agg(
    average_score=("score", "mean"),
    average_hours=("hours_studied", "mean"),
    students=("student", "count")
).round(2)
```

```python
df['score'].mean()
df['score'].median()
```

## Merge with another dataframe

Useful command: `merge()`.

```python
language_info = pd.DataFrame({
    "language": ["Python", "R", "Java"],
    "type": ["scripting", "statistical", "compiled"]
})
language_info
```

```python
df
```

```python
df.merge(language_info, on="language", how="left")
```

## Drop duplicates and rename columns

Useful commands: `drop_duplicates()`, `rename()`.

```python
df[["language"]].drop_duplicates()
```

```python
df[["language"]]
```

```python
df.rename(columns={"hours_studied": "hours", "passed": "exam_passed"})
```
